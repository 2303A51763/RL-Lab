```python

# Assignment 1: Value Iteration in Deterministic Grid World

# 1. Define parameters
gamma = 0.9
theta = 0.0001

normal_reward = -1
wall_reward = -1
goal_reward = 100

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

# 2. Define transition rules
# Grid:
# S0 | WALL
# S1 | G

transitions = {
    'S0': {
        'Up': ('S0', wall_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S0', wall_reward)
    },

    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', wall_reward),
        'Left': ('S1', wall_reward),
        'Right': ('G', goal_reward)
    }
}

# 3. Initialize state values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

print("INITIAL STATE VALUES")
print(V)

# 4. Display transition information
print("\nTRANSITION INFORMATION")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]

        print(state, action,
              "->", next_state,
              "Reward:", reward)


# 5. Value Iteration Algorithm

iteration = 0

while True:

    iteration += 1

    # Store previous state values
    old_V = V.copy()

    # Calculate new action values and state values
    Q = {}

    for state in states:

        Q[state] = {}

        for action in actions:

            next_state, reward = transitions[state][action]

            Q[state][action] = (
                reward + gamma * old_V[next_state]
            )

        # Update state value using maximum Q value
        V[state] = max(Q[state].values())

    # Calculate maximum change
    delta = max(
        abs(V[state] - old_V[state])
        for state in states
    )

    # Display iteration-wise values
    print(
        f"Iteration {iteration}: "
        f"S0 = {V['S0']:.4f}, "
        f"S1 = {V['S1']:.4f}, "
        f"Delta = {delta:.6f}"
    )

    # Check convergence
    if delta < theta:
        break


# 6. Calculate final Q values using optimal values

Q = {}

for state in states:

    Q[state] = {}

    for action in actions:

        next_state, reward = transitions[state][action]

        Q[state][action] = (
            reward + gamma * V[next_state]
        )


# 7. Find optimal policy

policy = {}

for state in states:

    best_action = max(
        Q[state],
        key=Q[state].get
    )

    policy[state] = best_action


# 8. Display optimal state values

print("\n========== OPTIMAL STATE VALUES ==========")

for state in states:
    print(f"V*({state}) = {V[state]:.4f}")

print("V*(G) = 0 (Terminal)")


# 9. Display final Q table

print("\n========== ACTION VALUE TABLE ==========")

print(f"{'State':<10}", end="")

for action in actions:
    print(f"{action:<12}", end="")

print()

for state in states:

    print(f"{state:<10}", end="")

    for action in actions:
        print(f"{Q[state][action]:<12.4f}", end="")

    print()


# 10. Display optimal policy

print("\n========== OPTIMAL POLICY ==========")

for state in states:
    print(f"{state} -> {policy[state]}")

print("G -> Terminal")


# 11. Bellman Optimality Verification

print("\n========== BELLMAN OPTIMALITY ==========")

for state in states:

    max_q = max(Q[state].values())

    print(
        f"{state}: V* = {V[state]:.4f}, "
        f"Max Q = {max_q:.4f}"
    )
```
