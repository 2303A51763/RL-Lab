```python

# Assignment 1: Policy Iteration in a Deterministic Grid World

# 1. Define parameters
gamma = 0.9
theta = 0.0001

normal_reward = -1
wall_reward = -1
goal_reward = 100

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

# 2. Define transitions
# Grid:
# S0 | WALL
# S1 | G

transitions = {
    'S0': {
        'Up': ('S0', wall_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S0', wall_reward)
    },

    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', wall_reward),
        'Left': ('S1', wall_reward),
        'Right': ('G', goal_reward)
    }
}

# 3. Initialize policy and state values
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

print("INITIAL POLICY:", policy)
print("INITIAL STATE VALUES:", V)

# 4. Policy Evaluation
def policy_evaluation(policy, V):

    iteration = 0

    while True:
        delta = 0
        old_V = V.copy()

        for state in states:

            action = policy[state]

            next_state, reward = transitions[state][action]

            new_value = reward + gamma * old_V[next_state]

            delta = max(delta, abs(new_value - old_V[state]))

            V[state] = new_value

        iteration += 1

        print(f"Evaluation {iteration}: "
              f"S0 = {V['S0']:.4f}, "
              f"S1 = {V['S1']:.4f}, "
              f"Delta = {delta:.6f}")

        if delta < theta:
            break

    return V


# 5. Calculate action values Q(s,a)
def calculate_Q(V):

    Q = {}

    for state in states:

        Q[state] = {}

        for action in actions:

            next_state, reward = transitions[state][action]

            Q[state][action] = (
                reward + gamma * V[next_state]
            )

    return Q


# 6. Policy Improvement
def policy_improvement(policy, Q):

    new_policy = {}

    for state in states:

        best_action = max(
            actions,
            key=lambda action: Q[state][action]
        )

        new_policy[state] = best_action

    return new_policy


# 7. Policy Iteration
policy_iteration = 0

while True:

    policy_iteration += 1

    print("\nPOLICY ITERATION:", policy_iteration)

    # Policy Evaluation
    V = policy_evaluation(policy, V)

    # Calculate Q values
    Q = calculate_Q(V)

    # Policy Improvement
    new_policy = policy_improvement(policy, Q)

    print("Old Policy:", policy)
    print("Improved Policy:", new_policy)

    # Check policy stability
    if new_policy == policy:

        print("Policy is stable!")
        break

    policy = new_policy


# 8. Final Results
print("\n========== FINAL RESULTS ==========")

print("\nOptimal State Values:")

for state in states:
    print(f"V*({state}) = {V[state]:.4f}")

print(f"V*(G) = {V['G']:.4f} (Terminal)")


# 9. Final Action-Value Table
Q = calculate_Q(V)

print("\nAction-Value Table Q*(s,a):")

print(f"{'State':<10}", end="")

for action in actions:
    print(f"{action:<12}", end="")

print()

for state in states:

    print(f"{state:<10}", end="")

    for action in actions:
        print(f"{Q[state][action]:<12.4f}", end="")

    print()


# 10. Optimal Policy
print("\nOptimal Policy:")

for state in states:
    print(f"{state} -> {policy[state]}")

print("G -> Terminal")


# 11. Bellman Optimality Verification
print("\nBellman Optimality Verification:")

for state in states:

    max_q = max(Q[state].values())

    print(
        f"{state}: V* = {V[state]:.4f}, "
        f"Max Q = {max_q:.4f}"
    )
```
