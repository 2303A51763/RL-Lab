```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 10

reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

print("===== EPSILON-GREEDY MULTI-ARMED BANDIT =====")
print("Epsilon =", epsilon)
print("Total Plays =", plays)
for i in range(plays):

    r = random.random()

    if r < epsilon:
        machine = random.choice(machines)
        method = "Exploration"
    else:
        machine = max(Q, key=Q.get)
        method = "Exploitation"

    reward = reward_pattern[machine][N[machine] % 4]

    N[machine] += 1
    R[machine].append(reward)

    # Incremental average
    Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

    total_reward += reward
    average_reward = total_reward / (i + 1)

    print("\n--------------------------------")
    print("Play Time:", i + 1)
    print("Random Value:", round(r, 3))
    print("Method:", method)
    print("Selected Machine:", machine)
    print("Reward:", reward)
    print("Machine Count N:", N[machine])
    print("Updated Q Average:", round(Q[machine], 2))
    print("Overall Average Reward:", round(average_reward, 2))

print("\n========== FINAL RESULTS ==========")

for machine in machines:
    print("\nMachine:", machine)
    print("Number of Plays N =", N[machine])
    print("Rewards =", R[machine])
    print("Average Q =", round(Q[machine], 2))

print("\n===================================")
print("Total Reward =", total_reward)
print("Overall Average Reward =", round(total_reward / plays, 2))

best = max(Q, key=Q.get)

print("Best Machine =", best)
print("Best Q Value =", round(Q[best], 2))
```
